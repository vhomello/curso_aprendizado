**Justificativa Pruning**
O método de pruning em árvores de regressão busca equilibrar a precisão do modelo e sua simplicidade, minimizando a seguinte função de erro
$$
\left(\sum^{|T|}_{m=1}\sum_{i:x_i \in R} (y_i - \hat{y_R})^2\right) + \alpha |T|
$$
Sendo
- $|T|$ o número de nós terminais
- $R$ a região asociada ao m-ésimo nó terminal
- $\hat{y}$ é a média dos target em R
- $\alpha$ parametro que penaliza arvores mais complexas

Ou seja, somamos o erro quadratico de cada região e e adiciona uma penalização proporcional ao número de nós terminais <br>
Dessa forma, $\alpha$ é proporcional ao viés e inversamente proporcional a variância


**Justificativa Número de árvores**
Uma maneira natural de reduzir a variância é gerar muitos subconjuntos de treinamento. Notar que, a média de
várias observações reduz a variância. Exemplo, considere um conjunto de n observações independentes $Z1, . . . , Zn$, cada uma com variância $\sigma^2$, a variância da média $\bar{Z}$ das observações é dada por $\sigma^2/n$.

Floresta Aleatória utiliza-se de Bootstrap aggregation (bagging) e bagging reduz a variância sem aumentar significativamente o viés
