```python
import numpy as np
import pandas as pd

from sklearn.linear_model import Ridge, Lasso, ElasticNet, LinearRegression
from sklearn.preprocessing import StandardScaler

# Reproducibility
np.random.seed(42)

# Generate synthetic data
n_samples = 100

# Independent variable
x1 = np.random.normal(0, 1, n_samples)

# Colinear variables: x2 ~ x1 + small noise
x2 = x1 + np.random.normal(0, 0.01, n_samples)

# Independent variable
x3 = np.random.normal(0, 1, n_samples)

# Independent variable
x4 = np.random.normal(0, 1, n_samples)

# Response variable: some true coefficients for illustration
y = 3 * x1 + 3 * x2 + 1 * x3 + np.random.normal(0, 0.1, n_samples)

# Stack features
X = np.column_stack([x1, x2, x3, x4])

# Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Fit Ridge and Lasso
ridge = Ridge(alpha=1)
lasso = Lasso(alpha=0.5)
elastic_net = ElasticNet(alpha=1, l1_ratio=0.5)  # alpha = λ1 + λ2, l1_ratio = λ1 / (λ1 + λ2)
ols = LinearRegression()

ridge.fit(X_scaled, y)
lasso.fit(X_scaled, y)
elastic_net.fit(X_scaled, y)
ols.fit(X_scaled, y)

# Create coefficient summary
coefficients = pd.DataFrame({
    'Variable': ['x1', 'x2', 'x3', 'x4'],
    'OLS Coeff': ols.coef_,
    'Ridge Coeff': ridge.coef_,
    'Lasso Coeff': lasso.coef_,
    'Elastic Net Coeff': elastic_net.coef_
})

coefficients
```
